## Checking result using Otto's proposed criteria

```python
df["N_pix_eff"] = df["A"] / df["FWHM"]**2

K = 16

df["threshold"] = (
    18.0
    + 2 * K * np.log(df["N_pix_eff"])
)

df["Otto_detectable"] = (
    df["SNR2"] > df["threshold"]
)

df_pass = df[df["Otto_detectable"]].copy()
```

```python
len(df_pass) == len(df)
```
